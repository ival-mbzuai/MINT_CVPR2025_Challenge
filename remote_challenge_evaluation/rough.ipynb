```python
# Function to extract unique IDs from the JSON
def extract_unique_ids(json_data):
    oe_ids = [entry['unique_id'] for entry in json_data['oe']]
    mcq_ids = [entry['unique_id'] for entry in json_data['mcq']]
    return oe_ids + mcq_ids  # Return combined list of unique IDs


root_dir_vol = "C:/Users/sanji/Desktop/Merciful/Multilingual_Dataset/Dataset/Volunteer_Verification/eval_ai_github/MINT_CVPR2025_Challenge"

# submission_file_path = download(submission, save_dir)
submission_file_path = root_dir_vol + "/annotations/submission_val_rough_3.json"
val_annotation_file_path = root_dir_vol + "/annotations/combined_GT_val_rough_3.json"
# test_annotation_file_path = root_dir_vol + "/annotations/combined_GT_test.json"

with open(submission_file_path, "r", encoding="utf-8") as file:
    data_sub = json.load(file)

with open(val_annotation_file_path, "r", encoding="utf-8") as file:
    data_val = json.load(file)

# Extract unique IDs from both JSONs
val_json_ids = extract_unique_ids(data_val)
sub_json_ids = extract_unique_ids(data_sub)

# Check if all unique_ids from second JSON are in the first JSON
missing_ids = set(sub_json_ids) - set(val_json_ids)
if missing_ids:
    print("Missing unique_ids from second JSON:", missing_ids)
else:
    print("No unique_ids are missing in the second JSON.")
```

```python
# Function to remove 'answer' and 'answer_choice' from JSON data
def clean_json(data):
    # Remove 'answer' from 'oe' section
    for item in data["oe"]:
        if "answer" in item:
            del item["answer"]
    
    # Remove 'answer' and 'answer_choice' from 'mcq' section
    for item in data["mcq"]:
        if "answer" in item:
            del item["answer"]
        if "answer_choice" in item:
            del item["answer_choice"]
    
    return data


root_dir_vol = "C:/Users/sanji/Desktop/Merciful/Multilingual_Dataset/Dataset/Volunteer_Verification"

test_annotation_file_path = root_dir_vol + "/CVPR_Workshop/json_files/combined_test.json"

with open(test_annotation_file_path, "r", encoding="utf-8") as file:
    data_test = json.load(file)


# Clean the JSON data
cleaned_data = clean_json(data_test)

# Save the cleaned data to a file
with open(root_dir_vol + '/CVPR_Workshop/json_files/combined_test_redacted.json', 'w', encoding='utf-8') as f:
    json.dump(cleaned_data, f, indent=4, ensure_ascii=False)
```

### Video shift

```python
import json
import shutil
import os

# Function to copy video files
def copy_videos(data, source, destination):
    for category in ['oe', 'mcq']:
        for item in data[category]:
            video_id = item['video_id']
            source_video_path = os.path.join(source, video_id)
            
            # Check if the video exists in the source directory
            if os.path.exists(source_video_path):
                destination_video_path = os.path.join(destination, video_id)
                
                # Copy the video file to the destination directory
                shutil.copy(source_video_path, destination_video_path)
                print(f"Copied {video_id} to {destination}")
            else:
                print(f"Video {video_id} not found in source directory")

root_dir_vol = "C:/Users/sanji/Desktop/Merciful/Multilingual_Dataset/Dataset/Volunteer_Verification"

set_list = ["val", "test"]

for temp_set in set_list:

    # Paths for the source and destination directories
    source_dir = 'I:/MINT/mint_dataset/'  # Folder where original videos are stored
    if temp_set == "val":
        destination_dir = 'I:/MINT/val_videos/'  # Folder to copy videos to
        json_file_path = root_dir_vol + "/CVPR_Workshop/json_files/combined_val.json"
    else:
        destination_dir = 'I:/MINT/test_videos/'  # Folder to copy videos to
        json_file_path = root_dir_vol + "/CVPR_Workshop/json_files/combined_test_redacted.json"

    # Load the cleaned JSON file
    with open(json_file_path, 'r', encoding='utf-8') as f:
        cleaned_data = json.load(f)



    # Create the destination directory if it doesn't exist
    if not os.path.exists(destination_dir):
        os.makedirs(destination_dir)

    # Copy the videos based on the video_id values from the JSON file
    copy_videos(cleaned_data, source_dir, destination_dir)

    print("Video copying process completed.")
```
